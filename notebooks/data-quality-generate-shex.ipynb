## Extracting the Shape Expressions

This notebook shows how to extract Shape Expressions (ShEx) in order to assess the data quality. ShEx describe node constraints that can be tested against a RDF dataset or a public SPARQL Endpoint. 

### First we load the python packages libraries 

````python
from shexer.shaper import Shaper
from shexer.consts import TURTLE, SHEXC, SHACL_TURTLE
````

### Then we select the classes that we want to analyse

````python
target_classes = [
    "https://schema.org/CreativeWork",
    "https://schema.org/Person"
]
````

### We define the namespaces used in our dataset 

````python
namespaces_dict = {"https://schema.org/": "schema",
                   "http://www.w3.org/2000/01/rdf-schema#": "rdfs", 
                   "https://example.org/": "ex",
                   "http://weso.es/shapes/": "",
                   "http://www.w3.org/2001/XMLSchema#": "xsd",
                   "http://schema.org/": "schema",
                   "http://rdfs.org/ns/void#": "void",
                   "http://creativecommons.org/ns#": "creativeCommons",
                   "http://purl.org/dc/terms/": "dc",
                   "http://purl.org/ontology/bibo/": "bibo",
                   "http://www.w3.org/2004/02/skos/core#": "skos"
}
````

````python
# We define the name of the input folder and file
folder_name = 'chile'
file_name = 'chile'

input_nt_file = "../datos/output/"+folder_name+"/dataset_"+file_name+".ttl"
````

### The Shaper class

The Shaper class is used to create the ShEx. We can provide the dataset as an input file, the format and the number of instances to be retrieved.

````python
shaper = Shaper(target_classes=target_classes,
                #raw_graph=raw_graph,
                graph_file_input=input_nt_file,
                #url_endpoint=url_endpoint, 
                #input_format=NT,
                input_format=TURTLE, 
                limit_remote_instances=20,
                namespaces_dict=namespaces_dict,  # Default: no prefixes
                instantiation_property="http://www.w3.org/1999/02/22-rdf-syntax-ns#type")  # Default rdf:type
````

**Note:** Threshold of tolerance. The constraints inferred for each shape may not be compatible with every node associated to the shapes. With this threshold you can indicate the minimum percentage of nodes that should conform with a constraint c. If c does not reach the indicated ratio, its associated information will not appear in the final shape. 

````python
output_file = "../datos/output/"+folder_name+"/"+file_name+"_shaper_elit_all.shex"

shaper.shex_graph(output_file=output_file,
                  acceptance_threshold=0.7)
````
